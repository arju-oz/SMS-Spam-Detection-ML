```python
combined_df.drop(['S. No.'],axis=1,inplace=True)
```

```python
combined_df['Label']=combined_df['Label'].map({'Non-Spam':0,'Spam':1})
combined_df
```

```python
nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
tweets=combined_df.Message_body
tweets
```

```python
#Tokenization
from nltk.tokenize import TweetTokenizer
tk=TweetTokenizer()
tweets=tweets.apply(lambda x: tk.tokenize(x)).apply(lambda x:" ".join(x))
tweets
```

```python
#Remove Special Characters
tweets=tweets.str.replace('[^a-zA-Z0-9]',' ',regex=True)
tweets
```

```python
from nltk.tokenize import word_tokenize
tweets=tweets.apply(lambda x: ' '.join([w for w in word_tokenize(x) if len(w)>=3]))
tweets
```

```python
#Stemming
from nltk.stem import SnowballStemmer
stemmer=SnowballStemmer('english')
twets=tweets.apply(lambda x:[stemmer.stem(i.lower()) for i in word_tokenize(x)]).apply(lambda x: ' '.join(x))
tweets
```

```python
#Remove stopwords
from nltk.corpus import stopwords
stop=stopwords.words('english')
tweets=tweets.apply(lambda x:[i for i in word_tokenize(x) if i not in stop]).apply(lambda x: ' '.join(x))
tweets
```

```python
#Vectorization
from sklearn.feature_extraction.text import TfidfVectorizer
vec=TfidfVectorizer()
train_data=vec.fit_transform(tweets)
```

```python
train_data.shape
```

```python
print(train_data)
```

```python
y=combined_df['Label'].values
y
```

```python
#Machine Learning mode
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(train_data,y,test_size=0.30,random_state=1)
x_train
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
knn=KNeighborsClassifier(n_neighbors=7)
sv=SVC()
base=MultinomialNB()
dec=DecisionTreeClassifier(criterion='entropy')
from sklearn.metrics import accuracy_score
```

```python
lst=[knn,sv,base,dec]
for i in lst:
  print("Model name is",i)
  i.fit(x_train,y_train)
  y_pred=i.predict(x_test)
  print(y_pred)
  print("*"*100)
  print("Accuracy score is",accuracy_score(y_test,y_pred))
```

```python
model=DecisionTreeClassifier(criterion='entropy')
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
y_pred
```
